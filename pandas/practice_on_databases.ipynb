## Filtering, Adding New Columns and Other Functions

### Filtering Rows and Columns

```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Rahul", "Akshita", "Aman", "Mohit", "Diksha", "Nonny"],
    "Age": [20, 18, 19,18,19,20],
    "Marks": [85, 90, 75, 60, 84, 74],
    "Branch": ["ME","CS", "CS","EE","CS","CS"]
})
```

```python
df.loc[[2,1], ["Name", "Age"]]
```

```python
df.sort_values(["Branch", "Marks", "Age"], ascending=[True, True, False])
```

```python
df[df["Marks"]>80]
```

```python
df[df["Branch"]=="CS"]
```

```python
df[(df["Branch"]== "CS")& (df["Marks"]> 80)]
```

```python
df[(df["Branch"]== "CS") | (df["Marks"]> 80)]
```

```python
df[~(df["Branch"]== "CS") & ~(df["Marks"]> 80)]  ## '~' means not
```

### Special Filtering Methods

```python
# isin()
df[df["Age"].isin([18,20])]
```

```python
## between()
df[df["Age"].between(18,20)]
```

### Adding and Modifying Columns

```python
df["City"] = ["Alwar", "Jaipur", "Shahpura DC", "Alwar", "Jaipur", "St. Bagar"]
```

```python
df["Percentage"] = (df["Marks"] + 500) / 600 * 100
```

```python
df
```

```python
df["Marks"] = df["Marks"] - 10
df
```

### Conditional Columns

```python
import numpy as np

df["Category"] = np.where(
    df["Marks"] >= 60,
    "Excellent",
    "Good"
)
```

```python
df
```

```python
conditions = [
    df["Marks"] >= 60,
    df["Marks"] >= 45,
    df["Marks"] >= 30
]

choices = [
    "A",
    "B",
    "C"
]

df["Grade"] = np.select(
    conditions,
    choices,
    default="F"
)

df
```

### Filtering Rows by Category

```python
## groupby()
df.groupby("Branch")["Marks"].mean()
```

```python
df.groupby("Branch")["Name"].count()
```

```python
### Multiple Aggregations
df.groupby("Branch").agg({"Marks": ["mean", "max", "min"],"Name":"count"})
```

### Merging Two Dataframes

```python
df2 = pd.DataFrame({
    "Name": ["Rahul", "Akshita", "Aman", "Mohit", "Diksha", "Nonny"],
    "Section": ["M1", "CA1", "CA1", "E1", "CY", "CB1"]
})
```

```python
df3 = pd.merge(df,df2, on = "Name" )
df3
```

```python
pd.concat([df,df2]) ## Vertical Concatenation
```

```python
new__column = pd.DataFrame({
    "College": ["GECA","GECA","GECA",np.nan,"GECA", "GECA"]
})
df = pd.concat([df,new__column], axis = 1) ## Vertical Concatenation
df
```

```python
new_row = pd.DataFrame({
    "Name": ["Rohit"],
    "Age" : [19],
    "Branch": ["CS"],
    "City": ["Madar DC"]
})
cf = pd.concat([df,new_row], ignore_index= True)
cf
```
