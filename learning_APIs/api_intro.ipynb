## URL Building with Parameters

📒 API Notes – Query Parameters & URL Building
1. What is a Query Parameter?

Extra information sent to the server along with the URL.

Syntax:

https://website.com/resource?key=value

Example:

https://example.com/profile?name=Nishant
? → Starts query parameters.
= → Assigns a value to a key.
& → Separates multiple parameters.
2. Multiple Query Parameters

Example:

https://example.com/profile?name=Nishant&age=20

Here:

Key	Value
name	Nishant
age	20
3. Using params (Recommended)

Instead of manually creating the URL:

❌

url = "https://example.com/profile?name=Nishant&age=20"

Use:

params = {
    "name": "Nishant",
    "age": 20
}

response = requests.get(url, params=params)

requests automatically creates:

https://example.com/profile?name=Nishant&age=20
4. Why Use params?

✅ Cleaner code

✅ No string concatenation

✅ Automatically handles URL encoding

Example:

params = {
    "city": "New York"
}

Automatically becomes:

?city=New+York

or

?city=New%20York

No manual work required.

5. Flow
Dictionary
      │
      ▼
params = {
    "name": "Nishant",
    "age": 20
}
      │
      ▼
requests.get(url, params=params)
      │
      ▼
https://example.com/profile?name=Nishant&age=20
6. General Pattern
url = "https://api.example.com"

params = {
    "key1": value1,
    "key2": value2
}

Query parameters allow us to send extra information to an API, and the params dictionary lets the requests library safely build the final URL automatically.|
